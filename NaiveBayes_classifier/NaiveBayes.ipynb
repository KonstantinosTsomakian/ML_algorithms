```python
import pandas as pd
from NaiveBayes import *
```

```python
path_to_data_dir = './example_data/'
```

Load the data

```python
categorical_data = pd.read_csv(f'{path_to_data_dir}/DatasetA_X_categorical.csv')
categorical_targets = pd.read_csv(f'{path_to_data_dir}/DatasetA_Y.csv')

continuous_data = pd.read_csv(f'{path_to_data_dir}/DatasetB_X_continuous.csv', header=None)
continuous_targets = pd.read_csv(f'{path_to_data_dir}/DatasetB_Y.csv', header=None)
```

Naive bayes algorithm is based on:
* The Bayes theorem
* The assumption that all the explanatory varibales(features) are independednt.

$$
P(C \mid X)
=
\frac{P(X \mid C)\,P(C)}{P(X)}
$$

The goal is to compute the probability of each class given the features. The most probable class given the features is the final prediction of the model.

____

```python
feature_type = 'categorical'

if feature_type == 'categorical':

	X = categorical_data
	Y = categorical_targets
elif feature_type == 'continuous':

	X = continuous_targets
	Y = continuous_targets




predictions, acc = NaiveBayes(X = X, Y = Y, training_percentage=0.8, feature_type=feature_type)

print(f'Model predicted {acc}% correct on the test set.')
```
